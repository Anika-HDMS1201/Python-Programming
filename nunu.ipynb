```python
student1={
    "Name":"Anika Jana",
    "Age": 23
        }

for student_id,student_info in student1.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
numbers=[1,2,2,3,3,3,4,4,4,4,]
frequency={}

for i in numbers:
    if i in frequency:
        frequency[i]+=1
    else:
        frequency[i]=1
print(frequency)
```

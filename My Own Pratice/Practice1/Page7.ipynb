```python
def even(num):
    if num%2==0:
        return True
even(24)
```

```python
lst=[1,2,3,4,5,6,7,8,10,11,12]
list(filter(even,lst))
```

```python
people=[
     {'Name':'Krish','Age':32},
     {'Name':'Jack','Age':35},
     {'Name':'John','Age':25}
]
def age_greater_than_25(person):
    return person['Age']>25

list(filter(age_greater_than_25,people))
```
